## Sprint 2 (Transformação de Strings, Integer e Float e Datetime)

```python
# Necessário remover colunas vazias geradas pelos ; do final da linha
df = df.dropna(how="all", axis=1)

df.head()
```

```python
# Limpeza de Strings e Remoção de Caracteres Especiais
cols_texto = ['PR_CAT', 'PR_NOME', 'CL_GENERO', 'CL_SEG']
for col in cols_texto:
    df[col] = df[col].astype(str).str.strip()
    df[col] = df[col].str.replace(r'[^\w\s-]', '', regex=True)
```

```python
# Limpeza de Inteiros
cols_inteiros = ['CO_ID', 'CL_ID', 'CL_EC', 'CL_FHL', 'PR_ID']
for col in cols_inteiros:
    df[col] = pd.to_numeric(df[col].astype(str).str.replace(r'\D', '', regex=True), errors='coerce')
```

```python
# Conversão da coluna DATA para datetime
df['DATA'] = pd.to_datetime(df['DATA'], dayfirst=True, errors='coerce')
```

## Sprint 3 (Limpeza de Nulos e Duplicatas):

```python
# 1. Preenchimento de categorias ausentes
df['PR_CAT'] = df['PR_CAT'].apply(
    lambda categoria: 'Sem Categoria' if categoria == 'ND' else categoria
)
```

```python
# 2. Tratamento do número de filhos
df['CL_FHL'] = df['CL_FHL'].apply(
    lambda x: 0 if pd.isna(x) else x
)
```

```python
# Agrupamento 1: Total de Itens Vendidos e Compras Únicas por Categoria
vendas_categoria = df.groupby('PR_CAT').agg(
    Qtd_Itens=('PR_ID', 'count'),
    Qtd_Notas_Unicas=('CO_ID', 'nunique')
).reset_index()

print("--- Agrupamento por Categoria ---\n")
print(vendas_categoria)
```

```python
# Agrupamento 2: Vendas por Segmentação Econômica do Cliente
vendas_segmento = df.groupby('CL_SEG').agg(
    Qtd_Itens=('PR_ID', 'count'),
    Qtd_Clientes_Unicos=('CL_ID', 'nunique')
).reset_index()

print("\n--- Agrupamento por Segmento do Cliente ---")
print(vendas_segmento)
```
